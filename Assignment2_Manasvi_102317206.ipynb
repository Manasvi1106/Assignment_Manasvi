```python
L = [10, 20, 30, 40, 50, 60, 70, 80]
L.append(200)
L.append(300)
print("New List:", L)
L.remove(10)
L.remove(30)
print("New List:", L)
L.sort()
print("Ascending order:", L)
L.sort(reverse=True)
print("Descending order:", L)
```

```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)
highest_score = max(scores)
highest_index = scores.index(highest_score)
print("Highest score:", highest_score, "at index:", highest_index)

lowest_score = min(scores)
lowest_count = scores.count(lowest_score)
print("Lowest score:", lowest_score, "Times appeared:", lowest_count)

reversed_list = list(scores[::-1])
print("Reversed tuple:", reversed_list)

specific_score = input("Enter a score: ")
if specific_score in scores:
    first_occurrence = scores.index(specific_score)
    print("Present at index:", first_occurrence)
else:
    print("Not present in the tuple")
```

```python
import random
numbers = [random.randint(100, 900) for _ in range(100)]

odd_numbers = [num for num in numbers if num % 2 != 0]
print("Odd numbers:", odd_numbers)
print("Total odd numbers:", len(odd_numbers))

even_numbers = [num for num in numbers if num % 2 == 0]
print("Even numbers:", even_numbers)
print("Total even numbers:", len(even_numbers))

prime_numbers = [
    num for num in numbers
    if num > 1 and all(num % i != 0 for i in range(2, int(num**0.5) + 1))]
print("Prime numbers:", prime_numbers)
print("Total prime numbers:", len(prime_numbers))

```

```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

unique = A | B
print("Unique scores:", unique)

common = A & B
print("Common scores:", common)

exclusive = A ^ B
print("Exclusive scores:", exclusive)

is_subset = A.issubset(B)
is_superset = B.issuperset(A)
print("scores of team A are a subset of team B: ",is_subset)
print("Team B's scores are a superset of team A: ",is_superset)

X = int(input("Enter a score: "))
if X in A:
    A.remove(X)
    print("New set A:", A)
else:
    print("Not present in set A.")
```
