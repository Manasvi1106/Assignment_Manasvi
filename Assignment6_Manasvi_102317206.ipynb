```python
#Q1
import numpy as np
import matplotlib.pyplot as plt
M = float(input("Enter a value for M: "))
x = np.linspace(-10, 10, 100)
y1 = M * x**2
y2 = M * np.sin(x)
plt.plot(x, y1,'r--',label=f'y = {M} * x^2')
plt.plot(x, y2,'b-',label=f'y = {M} * sin(x)')
plt.title('Mathematical Functions')
plt.legend()
plt.grid(True)
plt.show()
```

```python
##Q2
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = {
    'Subject': ['Maths', 'Physics', 'English', 'Chemistry', 'Hindi'],
    'Score': [95, 90, 92, 93, 78]
}
df = pd.DataFrame(data)
colors = ['red', 'blue', 'green', 'pink', 'yellow']
plt.figure(figsize=(8, 6))
bars = plt.bar(df['Subject'], df['Score'], color=colors)
plt.text(0, 95+1, 'Maths', ha='center', color='black')
plt.text(1, 90+1, 'Physics', ha='center', color='black')
plt.text(2, 92+1, 'English', ha='center', color='black')
plt.text(3, 93+1, 'Chemistry', ha='center', color='black')
plt.text(4, 78+1, 'Hindi', ha='center', color='black')
plt.title('Scores of Different Subjects')
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.grid()
plt.show()


```

```python
##Q3
import numpy as np
import matplotlib.pyplot as plt
roll_number = int(input("Enter your roll number: "))
np.random.seed(roll_number)
data = np.random.rand(50)
print(data)
fig, ax = plt.subplots(1, 2)
ax[0].plot(np.cumsum(data))
ax[0].set_title('Cumulative Sum')
ax[0].set_xlabel('Index')
ax[0].set_ylabel('Cumulative Sum')
ax[0].grid(True)
ax[1].scatter(range(len(data)), data, color='r')
ax[1].set_title('Random Noise')
ax[1].set_xlabel('Index')
ax[1].set_ylabel('Random Noise')
ax[1].grid(True)
plt.tight_layout()
plt.show()
```

```python
##Q4
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv(
    "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
)

plt.figure(figsize=(10, 6))
plt.plot(
    data["month_number"], data["total_profit"], color="blue", marker="o", linestyle="-"
)
plt.title("Total Profit per Month")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
for column in data.columns[1:-1]:
    plt.plot(data["month_number"], data[column], label=column)

plt.title("Product Sales Data")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()

data.plot(kind="bar", figsize=(15, 8))
plt.title("Company Sales Data")
plt.xlabel("Month")
plt.ylabel("Values")
plt.grid(True, linestyle="--", alpha=0.7)
plt.show()
data = pd.read_csv(
    "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
)

plt.figure(figsize=(10, 6))
plt.plot(
    data["month_number"], data["total_profit"], color="blue", marker="o", linestyle="-"
)
plt.title("Total Profit per Month")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
for column in data.columns[1:-1]:
    plt.plot(data["month_number"], data[column], label=column)

plt.title("Product Sales Data")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()

data.plot(kind="bar", figsize=(15, 8))
plt.title("Company Sales Data")
plt.xlabel("Month")
plt.ylabel("Values")
plt.grid(True, linestyle="--", alpha=0.7)
plt.show()
```
