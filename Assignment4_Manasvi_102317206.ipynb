```python
##Q2
import numpy as np
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print(reversed_arr)
x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
def most_frequent(arr):
  values, counts = np.unique(arr, return_counts=True)
  max_count_index = np.argmax(counts)
  most_frequent_value = values[max_count_index]
  indices = np.where(arr == most_frequent_value)[0]
  return most_frequent_value, indices
most_frequent_x, indices_x = most_frequent(x)
print("Most frequent value in x: ",most_frequent_x, "has indices: ",indices_x)
most_frequent_y, indices_y = most_frequent(y)
print("Most frequent value in y: ",most_frequent_y, "has indices: ",indices_y)
```

```python
##Q4
import numpy as np
Manasvi = np.linspace(10, 100, 25)
print("Array:", Manasvi)
print("Dimensions of the array:", Manasvi.ndim)
print("Shape of the array:", Manasvi.shape)
print("Total elements in the array:", Manasvi.size)
print("Data type of each element:", Manasvi.dtype)
print("Total bytes consumed by the array:", Manasvi.nbytes)
Manasvi_transposed = Manasvi.reshape(25, 1)
print("Transposed array: ", Manasvi_transposed)
print("Transposed array using T attribute: ", Manasvi.T)
```

```python
##Q5
import numpy as np
UCS420_Manasvi = np.array([[10, 20, 30, 40],
                            [50, 60, 70, 80],
                            [90, 15, 20, 35]])
print("Array:\n", UCS420_Manasvi)
print("Mean:", np.mean(UCS420_Manasvi))
print("Median:", np.median(UCS420_Manasvi))
print("Maximum value:", np.max(UCS420_Manasvi))
print("Minimum value:", np.min(UCS420_Manasvi))
print("Unique elements:", np.unique(UCS420_Manasvi))
reshaped = UCS420_Manasvi.reshape(4, 3)
print("Reshaped Array: ", reshaped)
resized = np.resize(UCS420_Manasvi, (2, 3))
print("Resized Array: ", resized)
```
