```python
import random
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

num = random.randint(1, 100)
print(f"{num} is prime.") if is_prime(num) else print(f"{num} is not prime.")
```
